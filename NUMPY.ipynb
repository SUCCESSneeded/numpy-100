How to Add a Border Around a NumPy Array?

```python
'''| Mode          | Meaning                         | Example                       |
| ------------- | ------------------------------- | ----------------------------- |
| `constant`    | Adds a fixed value              | `[1,2,3] → [0,1,2,3,0]`       |
| `edge`        | Repeats edge values             | `[1,2,3] → [1,1,2,3,3]`       |
| `linear_ramp` | Adds values in a linear ramp    | gradually increases/decreases |
| `maximum`     | Pads using maximum value        | `[1,2,3] → [3,1,2,3,3]`       |
| `minimum`     | Pads using minimum value        | `[1,2,3] → [1,1,2,3,1]`       |
| `mean`        | Pads using mean                 | mean of array/edge            |
| `median`      | Pads using median               | median of array/edge          |
| `wrap`        | Wraps values from opposite side | `[1,2,3] → [3,1,2,3,1]`       |
| `symmetric`   | Mirrors values including edge   | `[1,2,3] → [2,1,2,3,2]`       |
| `reflect`     | Mirrors values excluding edge   | `[1,2,3] → [2,1,2,3,2]`       |
'''
```

Fibonacci Series using Binet Formula

```python
import numpy as np

# We are creating an array contains n = 10 elements
# for getting first 10 Fibonacci numbers
a = np.arange(1, 11)
lengthA = len(a)

# splitting of terms for easiness
sqrtFive = np.sqrt(5)
alpha = (1 + sqrtFive) / 2
beta = (1 - sqrtFive) / 2

# Implementation of formula
# np.rint is used for rounding off to integer
Fn = np.rint(((alpha ** a) - (beta ** a)) / (sqrtFive))
print("The first {} numbers of Fibonacci series are {} . ".format(lengthA, Fn))
```

How to Calculate the Determinant of a Matrix using NumPy


```python
# numpy.linalg.slogdet(square matrix)

''' It computes the sign and the logarithm of the determinant separately, which helps prevent numerical overflow or underflow when dealing with
 very large or very small values.
'''

A = np.array([[50, 29], [30, 44]])
sign, logdet = np.linalg.slogdet(A)
res = sign * np.exp(logdet)
print(res)
```

```python
#scipy.linalg.lu()
import scipy.linalg
'''LU decomposition can also be used to calculate the determinant by decomposing the matrix into lower (L) and upper (U) triangular matrices.
 The determinant is the product of the diagonal elements of the U matrix.'''

A = np.array([[1, 2], [3, 4]])
P, L, U = scipy.linalg.lu(A)
res = np.prod(np.diag(U))
print(res)

'''Matrix
How to calculate it
U:U Use row operations to make the elements below the diagonal zero.

L:L Store the elimination multipliers below the diagonal; put 1s on the diagonal.

P:P Record row swaps; if none are needed, use the identity matrix.
'''
```

Convert Covariance Matrix to Correlation Matrix using Python

```python
import seaborn as sns
iris=sns.load_dataset("iris")
iris.head()
```

```python
data = iris.iloc[:, :-1].values
```

```python
def calCov(x,y):
  x_mean,y_mean=np.mean(x),np.mean(y)
  cov=np.mean((x-x_mean)*(y-y_mean))
  return cov
```

```python
def covMat(data):
  rows,cols=data.shape
  cov_matrix=np.zeros((cols,cols))
  for i in range(cols):
    for j in range(cols):
      cov_matrix[i][j]=calCov(data[:,i],data[:,j])

  return cov_matrix
```

```python
covMat(data)
```

```python
np.cov(data, rowvar=False) #rowvar=False ensures columns are treated as features and rows as observations.
```

Convert Covariance Matrix to Correlation Matrix

```python
#Convert Covariance Matrix to Correlation Matrix
def corMat(data):
  rows,cols=data.shape
  corr_matrix=np.zeros((cols,cols))
  for i in range(cols):
    for j in range(cols):
      x,y=data[:,i],data[:,j]
      corr_matrix[i][j]=calCov(x,y) / (np.std(x)*np.std(y))
  return corr_matrix

corMat(data)
```

Replace NaN Values with Average of Columns


```python
#np.nanmean(arr, axis=0)  -> mean of i=each columns ignoring nan
#np.take(col_mean, inds[1])  -> mean of columns corresponding to nan
arr = np.array([[1.3, 2.5, 3.6, np.nan], [2.6, 3.3, np.nan, 5.5], [2.1, 3.2, 5.4, np.nan]])
col_mean = np.nanmean(arr, axis=0)
inds = np.where(np.isnan(arr))
print(arr)
print(inds)
arr[inds] = np.take(col_mean, inds[1])
print(arr)

# inds[0] → array([0, 1, 2]) — row indices.
# inds[1] → array([3, 2, 3]) — column indices.
```

```python
'''Function	                                             Purpose
np.ma.array(data, mask=...)	                Creates a masked array
np.ma.masked_array(data, mask=...)	        Creates a masked array (similar to array)
np.ma.masked_where(condition, a)	          Masks elements where the condition is True
np.ma.masked_equal(x, value)	              Masks elements equal to a given value
np.ma.masked_greater(x, value)	            Masks elements greater than a value
np.ma.masked_less(x, value)	                Masks elements less than a value
np.ma.masked_invalid(a)	                    Masks NaN and infinite values
np.ma.getmask(a)	                          Returns the mask
np.ma.filled(a, fill_value)	                Returns an array with masked elements replaced
np.ma.mean(a, axis=None)	                  Calculates the mean while ignoring masked elements'''
```

```python
arr = np.array([[1.3, 2.5, 3.6, np.nan], [2.6, 3.3, np.nan, 5.5], [2.1, 3.2, 5.4, 6.5]])
res=np.where(np.isnan(arr), np.ma.array(arr, mask=np.isnan(arr)).mean(),arr)
print(res)
```

How to Add a Border Around a NumPy Array?

```python
'''| Mode          | Meaning                         | Example                       |
| ------------- | ------------------------------- | ----------------------------- |
| `constant`    | Adds a fixed value              | `[1,2,3] → [0,1,2,3,0]`       |
| `edge`        | Repeats edge values             | `[1,2,3] → [1,1,2,3,3]`       |
| `linear_ramp` | Adds values in a linear ramp    | gradually increases/decreases |
| `maximum`     | Pads using maximum value        | `[1,2,3] → [3,1,2,3,3]`       |
| `minimum`     | Pads using minimum value        | `[1,2,3] → [1,1,2,3,1]`       |
| `mean`        | Pads using mean                 | mean of array/edge            |
| `median`      | Pads using median               | median of array/edge          |
| `wrap`        | Wraps values from opposite side | `[1,2,3] → [3,1,2,3,1]`       |
| `symmetric`   | Mirrors values including edge   | `[1,2,3] → [2,1,2,3,2]`       |
| `reflect`     | Mirrors values excluding edge   | `[1,2,3] → [2,1,2,3,2]`       |
'''
```

```python
#USING numpy.pad(array, pad_width, mode='constant', constant_values=value)
```

```python
#border of -1 around a 4×1 column vector.
arr=np.array([[10],[20],[30],[40]])
padded=np.pad(arr,pad_width=1,mode="constant",constant_values=-1)
print(arr)
print(padded)
```

```python
#a border is added around a 2×2 array. A padding width of 1 and a border value of 0 are used to surround the original matrix.

arr=np.zeros((2,3))
padded2=np.pad(arr,pad_width=1,mode="constant",constant_values=1)
print(arr)
print(padded2)
```

```python
#pads only the top and bottom of an array by using different padding widths for each side.
arr = np.arange(6).reshape(2, 3)
out = np.pad(arr, pad_width=((1, 1), (0, 0)), mode="constant", constant_values=9)
print(out)
```

Fibonacci Series using Binet Formula

```python
import numpy as np

# We are creating an array contains n = 10 elements
# for getting first 10 Fibonacci numbers
a = np.arange(1, 11)
lengthA = len(a)

# splitting of terms for easiness
sqrtFive = np.sqrt(5)
alpha = (1 + sqrtFive) / 2
beta = (1 - sqrtFive) / 2

# Implementation of formula
# np.rint is used for rounding off to integer
Fn = np.rint(((alpha ** a) - (beta ** a)) / (sqrtFive))
print("The first {} numbers of Fibonacci series are {} . ".format(lengthA, Fn))
```

How to Calculate the Determinant of a Matrix using NumPy


```python
# numpy.linalg.slogdet(square matrix)

''' It computes the sign and the logarithm of the determinant separately, which helps prevent numerical overflow or underflow when dealing with
 very large or very small values.
'''

A = np.array([[50, 29], [30, 44]])
sign, logdet = np.linalg.slogdet(A)
res = sign * np.exp(logdet)
print(res)
```

```python
#np.linalg.det()

det=LA.det(A)
print(det)
```

```python
#scipy.linalg.lu()
import scipy.linalg
'''LU decomposition can also be used to calculate the determinant by decomposing the matrix into lower (L) and upper (U) triangular matrices.
 The determinant is the product of the diagonal elements of the U matrix.'''

A = np.array([[1, 2], [3, 4]])
P, L, U = scipy.linalg.lu(A)
res = np.prod(np.diag(U))
print(res)

'''Matrix
How to calculate it
U:U Use row operations to make the elements below the diagonal zero.

L:L Store the elimination multipliers below the diagonal; put 1s on the diagonal.

P:P Record row swaps; if none are needed, use the identity matrix.
'''
```

```python
# np.bincount(x, weights=None, minlength=0)
# minlength ensures that the output contains at least the specified number of elements.

arr = np.array([0, 2, 2, 1, 4, 0, 2])
unique = np.unique(arr)
counts = np.bincount(arr)
print("Unique Values:", unique)
print("Counts:", counts[unique])
```
